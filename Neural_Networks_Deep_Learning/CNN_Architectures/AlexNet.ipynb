```python
import torch.nn as nn
```

```python
class AlexNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Conv2d(kernel_size=11, stride=4, in_channels=3, out_channels=96, padding="valid"),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=3, stride=2),
            nn.Conv2d(kernel_size=5, stride=1, in_channels=96, out_channels=256, padding="same"),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=3, stride=2),
            nn.Conv2d(kernel_size=3, stride=1, in_channels=256, out_channels=384, padding="same"),
            nn.ReLU(),
            nn.Conv2d(kernel_size=3, stride=1, in_channels=384, out_channels=384, padding="same"),
            nn.ReLU(),
            nn.Conv2d(kernel_size=3, stride=1, in_channels=384, out_channels=256, padding="same"),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=3, stride=2),

            nn.Flatten(),
            nn.Linear(9216, 4096),
            nn.ReLU(),
            nn.Linear(4096, 4096),
            nn.ReLU(),
            nn.Linear(4096, 1000),
        )

    def forward(self, inputs):
        return self.layers(inputs)
```
