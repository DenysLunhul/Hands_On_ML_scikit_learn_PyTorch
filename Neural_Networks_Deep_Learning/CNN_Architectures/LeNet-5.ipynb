```python
import torch.nn as nn
```

```python
class LeNet5(nn.Module):
    def __init__(self):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Conv2d(kernel_size=5, padding="valid", in_channels=1, out_channels=6),
            nn.Tanh(),
            nn.AvgPool2d(kernel_size=2, stride=2),
            nn.Tanh(),
            nn.Conv2d(kernel_size=5, padding="valid", in_channels=6, out_channels=16),
            nn.Tanh(),
            nn.AvgPool2d(kernel_size=2, stride=2),
            nn.Tanh(),
            nn.Conv2d(kernel_size=5, padding="valid", in_channels=16, out_channels=120),

            nn.Flatten(),
            nn.Linear(120, 84),
            nn.Tanh(),
            nn.Linear(84, 10),
        )

    def forward(self, input):
        return self.layers(input)
```
