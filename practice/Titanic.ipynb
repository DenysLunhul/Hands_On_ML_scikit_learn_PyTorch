```python
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OneHotEncoder
from sklearn.pipeline import make_pipeline, Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
import numpy as np

drop_columns = ["Cabin", "Ticket", "Name", "PassengerId"]

log_features = ["Age", "Fare"]
num_features = ["SibSp", "Parch", "HasCabin", "FamilySize"]
cat_features = ["Sex", "Pclass", "Embarked"]

log_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    FunctionTransformer(np.log1p),
    StandardScaler()
)

num_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
)

def add_features(df):
    df = df.copy()
    df["HasCabin"] = df["Cabin"].notna().astype(int)
    df["FamilySize"] = df["Parch"] + df["SibSp"] + 1
    return df

def drop_row_columns(df):
    return df.drop(columns=drop_columns)

cat_pipeline = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)
```

```python
preprocessing = Pipeline([
    ("features", FunctionTransformer(add_features, validate=False)),
    ("drop_raw", FunctionTransformer(drop_row_columns, validate=False)),
    ("columns", ColumnTransformer([
        ("log", log_pipeline, log_features),
        ("num", num_pipeline, num_features),
        ("cat", cat_pipeline, cat_features),
    ]))
])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV


model = make_pipeline(
    preprocessing,
    RandomForestClassifier(random_state=42)
)

search_grid = {
    "randomforestclassifier__n_estimators": [200, 400],
    "randomforestclassifier__max_depth": [None, 6],
    "randomforestclassifier__min_samples_split": [2, 10],
    "randomforestclassifier__min_samples_leaf": [1, 4],
    "randomforestclassifier__bootstrap": [True],
    "randomforestclassifier__class_weight": [None, "balanced"],
}

grid_search = GridSearchCV(model, search_grid, cv=3, scoring="accuracy")
grid_search.fit(X_train, y_train)
tuned_model = grid_search.best_estimator_
```

```python
pred = tuned_model.predict(X_test)
res = pd.DataFrame({"PassengerId": X_test.PassengerId,
                    "Survived": pred})
res.to_csv("titanic_submission.csv", index=False)
```
